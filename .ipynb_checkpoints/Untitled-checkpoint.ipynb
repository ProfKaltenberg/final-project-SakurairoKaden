```python
import csv
import time
import re
import requests

FILE_NAME = "steam_games.csv"
STORE_URL = "https://store.steampowered.com/api/appdetails"


def is_missing(value):
    if value is None:
        return True

    value = str(value).strip()

    return value == "" or value.lower() == "none" or value.lower() == "nan"


def get_release_date(appid):
    params = {
        "appids": appid,
        "filters": "release_date"
    }

    try:
        response = requests.get(STORE_URL, params=params, timeout=10)

        if response.status_code == 429:
            print("Rate limited. Waiting...")
            time.sleep(30)
            return get_release_date(appid)

        data = response.json()
        app_data = data.get(str(appid), {})

        if not app_data.get("success"):
            return None

        release_info = app_data.get("data", {}).get("release_date", {})
        return release_info.get("date")

    except:
        return None


def get_release_year(release_date):
    if is_missing(release_date):
        return None

    match = re.search(r"\d{4}", release_date)

    if match:
        return int(match.group())

    return None


def update_missing_release_dates():
    with open(FILE_NAME, "r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)
        rows = list(reader)
        fieldnames = reader.fieldnames

    if "release_date" not in fieldnames:
        fieldnames.append("release_date")

    if "release_year" not in fieldnames:
        fieldnames.append("release_year")

    updated = 0

    for row in rows:
        appid = row.get("appid")

        if is_missing(row.get("release_date")):
            print("Getting release date for", row.get("name"))

            release_date = get_release_date(appid)
            row["release_date"] = release_date

            time.sleep(0.5)

        if is_missing(row.get("release_year")):
            row["release_year"] = get_release_year(row.get("release_date"))

        if not is_missing(row.get("release_date")) and not is_missing(row.get("release_year")):
            updated += 1

    with open(FILE_NAME, "w", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(file, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(rows)

    print("Done. Checked", len(rows), "games.")
    print("Rows with release date and year now:", updated)


update_missing_release_dates()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

FILE_NAME = "steam_games.csv"
CLEAN_FILE = "steam_games_clean.csv"

games = pd.read_csv('/Users/vickywu/Documents/ECO 590 Data Analytics/final-project-SakurairoKaden/steam_games.csv')

games.head()

print("Rows before cleaning:", len(games))

missing_dates = games[games["release_date"].isna() | games["release_year"].isna()]

print("Rows with missing release date or year:", len(missing_dates))

games = games.dropna(subset=["release_date", "release_year"])

games["release_year"] = pd.to_numeric(games["release_year"])
games["positive_ratio"] = pd.to_numeric(games["positive_ratio"])
games["total_reviews"] = pd.to_numeric(games["total_reviews"])

games = games.dropna(subset=["release_year", "positive_ratio", "total_reviews"])

games = games[games["total_reviews"] > 0]

games.head()

print("Rows after cleaning:", len(games))

games.to_csv(CLEAN_FILE, index=False)

print("Saved cleaned data to", CLEAN_FILE)
```
